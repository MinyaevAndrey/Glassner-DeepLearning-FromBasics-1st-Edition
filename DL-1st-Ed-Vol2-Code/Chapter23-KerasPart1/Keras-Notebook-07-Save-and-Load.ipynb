### Notebook 7: Save and load models

```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Input
from tensorflow.keras.utils import to_categorical  # Исправленный импорт
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import backend as keras_backend
keras_backend.set_image_data_format('channels_last')
```

```python
# Here's a little model just for demonstration purposes
# from keras.models import Sequential
# from keras.layers import Dense

def make_model():
    model = Sequential([Input(shape=(10,)),
                        Dense(10, activation='relu'),
                        Dense(5, activation='softmax')]
    )
    model.compile(loss='categorical_crossentropy',
                  optimizer='adam',
                  metrics=['accuracy']
    )
    return model
```

```python
model = make_model()
```

```python
# How to save the model and weights
# We assume that there's a directory named SavedModels in the same
# directory as this file. If there isn't, this will probably cause an error.
# model.save('One_layer_model.h5')
model.save('One_layer_model.keras')
```

```python
# How to load the model and weights
from keras.models import load_model

# model2 = load_model('One_layer_model.h5')
model2 = load_model('One_layer_model.keras')
```

```python
# How to save just the weights
model.save_weights('One_layer_model.weights.h5')
```

```python
#How to load just the weights into an existing model
model2 = make_model()
model2.load_weights('One_layer_model.weights.h5')
```

```python
# ============ СОХРАНЕНИЕ ============
# Сохраняем архитектуру как JSON файл
filename = 'one_layer_model_architecture.json'  # Используем .json расширение

json_string = model.to_json()  # На самом деле это JSON строка
with open(filename, 'w') as outfile:
    outfile.write(json_string)  # Просто пишем JSON строку

print(f"Архитектура сохранена в {filename}")

# ============ ЗАГРУЗКА ============
from keras.models import model_from_json

# Загружаем архитектуру
with open(filename, 'r') as json_file:
    loaded_model_json = json_file.read()

# Восстанавливаем модель
loaded_model = model_from_json(loaded_model_json)

# Компилируем модель (так как веса не загружены)
loaded_model.compile(loss='categorical_crossentropy',
                     optimizer='adam',
                     metrics=['accuracy'])

print("Модель успешно загружена!")
loaded_model.summary()
```
