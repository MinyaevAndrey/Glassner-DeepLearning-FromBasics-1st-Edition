```python
# First we'll build our deep network and train it on the MNIST data.
# Then we can turn it loose on the Seattle photographs. We'll train
# for 20 epochs.

from keras.metrics import F1Score

random_seed = 42

# load MNIST data and save sizes
(X_train, y_train), (X_test, y_test) = mnist.load_data()
image_height = X_train.shape[1]
image_width = X_train.shape[2]
number_of_pixels = image_height * image_width

# convert to floating-point
X_train = keras_backend.cast_to_floatx(X_train)
X_test = keras_backend.cast_to_floatx(X_test)

# scale data to range [0, 1]
X_train /= 255.0
X_test /= 255.0

# save the original y_train and y_test
original_y_train = y_train
original_y_test = y_test

# replace label data with one-hot encoded versions
number_of_classes = int(1 + max(np.append(y_train, y_test)))
y_train = to_categorical(y_train, num_classes=number_of_classes)
y_test = to_categorical(y_test, num_classes=number_of_classes)

# reshape samples to 2D grid, one line per image
X_train = X_train.reshape([X_train.shape[0], number_of_pixels])
X_test = X_test.reshape([X_test.shape[0], number_of_pixels])

def make_model():
    model = Sequential([Input(shape=(number_of_pixels,)),
                        Dense(number_of_pixels, activation='relu'),
                        Dense(number_of_classes, activation='softmax')]
    )
    model.compile(loss='categorical_crossentropy',
                  optimizer='adam',
                  metrics=['accuracy', 'precision', 'recall', F1Score(average='macro', threshold=0.5)])
    return model

# make the model
model = make_model()

np.random.seed(random_seed)

# call fit() to train the model, and save the history
one_hidden_layer_history = model.fit(X_train, y_train,
                                     validation_data=(X_test, y_test),
                                     epochs=10,
                                     batch_size=256,
                                     verbose=2
)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Извлекаем метрики из истории обучения
train_loss = one_hidden_layer_history.history['loss']
val_loss = one_hidden_layer_history.history['val_loss']
train_accuracy = one_hidden_layer_history.history['accuracy']
val_accuracy = one_hidden_layer_history.history['val_accuracy']
train_precision = one_hidden_layer_history.history['precision']
val_precision = one_hidden_layer_history.history['val_precision']
train_recall = one_hidden_layer_history.history['recall']
val_recall = one_hidden_layer_history.history['val_recall']
train_f1 = one_hidden_layer_history.history['f1_score']
val_f1 = one_hidden_layer_history.history['val_f1_score']
epochs = range(1, len(train_loss) + 1)

# Создаем 5 графиков: Loss, Accuracy, Precision, Recall, F1
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()  # Упрощаем индексацию

# График 1: Loss (ошибки)
axes[0].plot(epochs, train_loss, 'b-', label='Обучающая выборка', linewidth=2)
axes[0].plot(epochs, val_loss, 'r--', label='Тестовая выборка', linewidth=2)
axes[0].set_xlabel('Эпохи', fontsize=11)
axes[0].set_ylabel('Потери', fontsize=11)
axes[0].set_title('Функция потерь (Loss)', fontsize=13)
axes[0].legend(fontsize=9)
axes[0].grid(True, alpha=0.3)
axes[0].set_xticks(epochs)

# График 2: Accuracy (точность)
axes[1].plot(epochs, train_accuracy, 'b-', label='Обучающая выборка', linewidth=2)
axes[1].plot(epochs, val_accuracy, 'r--', label='Тестовая выборка', linewidth=2)
axes[1].set_xlabel('Эпохи', fontsize=11)
axes[1].set_ylabel('Accuracy', fontsize=11)
axes[1].set_title('Точность (Accuracy)', fontsize=13)
axes[1].legend(fontsize=9)
axes[1].grid(True, alpha=0.3)
axes[1].set_xticks(epochs)

# График 3: Precision
axes[2].plot(epochs, train_precision, 'b-', label='Обучающая выборка', linewidth=2)
axes[2].plot(epochs, val_precision, 'r--', label='Тестовая выборка', linewidth=2)
axes[2].set_xlabel('Эпохи', fontsize=11)
axes[2].set_ylabel('Precision', fontsize=11)
axes[2].set_title('Точность предсказания (Precision)', fontsize=13)
axes[2].legend(fontsize=9)
axes[2].grid(True, alpha=0.3)
axes[2].set_xticks(epochs)

# График 4: Recall
axes[3].plot(epochs, train_recall, 'b-', label='Обучающая выборка', linewidth=2)
axes[3].plot(epochs, val_recall, 'r--', label='Тестовая выборка', linewidth=2)
axes[3].set_xlabel('Эпохи', fontsize=11)
axes[3].set_ylabel('Recall', fontsize=11)
axes[3].set_title('Полнота (Recall)', fontsize=13)
axes[3].legend(fontsize=9)
axes[3].grid(True, alpha=0.3)
axes[3].set_xticks(epochs)

# График 5: F1-score
axes[4].plot(epochs, train_f1, 'b-', label='Обучающая выборка', linewidth=2)
axes[4].plot(epochs, val_f1, 'r--', label='Тестовая выборка', linewidth=2)
axes[4].set_xlabel('Эпохи', fontsize=11)
axes[4].set_ylabel('F1-score', fontsize=11)
axes[4].set_title('F1-мера (F1-score)', fontsize=13)
axes[4].legend(fontsize=9)
axes[4].grid(True, alpha=0.3)
axes[4].set_xticks(epochs)

# Убираем шестой (пустой) подграфик, так как у нас 5 графиков
axes[5].set_visible(False)

plt.tight_layout()
plt.show()

# Вывод финальных значений
print("="*60)
print("Финальные значения на обучающей выборке:")
print(f"Loss:      {train_loss[-1]:.4f}")
print(f"Accuracy:  {train_accuracy[-1]:.4f}")
print(f"Precision: {train_precision[-1]:.4f}")
print(f"Recall:    {train_recall[-1]:.4f}")
print(f"F1-score:  {train_f1[-1]:.4f}")
print("="*60)
print("Финальные значения на тестовой выборке:")
print(f"Loss:      {val_loss[-1]:.4f}")
print(f"Accuracy:  {val_accuracy[-1]:.4f}")
print(f"Precision: {val_precision[-1]:.4f}")
print(f"Recall:    {val_recall[-1]:.4f}")
print(f"F1-score:  {val_f1[-1]:.4f}")
print("="*60)
```
